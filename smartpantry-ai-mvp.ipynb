```python
import pandas as pd

inventory = pd.DataFrame({
    'ingredient': [
        'Chicken',
        'Rice',
        'Tomato',
        'Eggs',
        'Yogurt'
    ]
})

inventory
```

```python
recipes = pd.DataFrame({
    'recipe_name': [
        'Chicken Rice Bowl',
        'Tomato Omelette',
        'Yogurt Salad'
    ],
    'ingredients': [
        ['Chicken', 'Rice'],
        ['Tomato', 'Eggs'],
        ['Yogurt', 'Tomato']
    ],
    'calories': [
        520,
        280,
        180
    ],
    'protein': [
        42,
        18,
        8
    ]
})

recipes
```

```python
available = set(inventory['ingredient'])

recommended = []

for _, row in recipes.iterrows():
    recipe_ingredients = set(row['ingredients'])

    if recipe_ingredients.issubset(available):
        recommended.append(row)

recommended_df = pd.DataFrame(recommended)

recommended_df
```

```python
for _, recipe in recommended_df.iterrows():

    print("=" * 40)

    print(f"🍽 Recipe: {recipe['recipe_name']}")
    print(f"🔥 Calories: {recipe['calories']}")
    print(f"💪 Protein: {recipe['protein']} g")
```

```python
inventory_expiry = pd.DataFrame({
    'ingredient': [
        'Chicken',
        'Tomato',
        'Yogurt',
        'Rice'
    ],

    'days_left': [
        2,
        4,
        1,
        60
    ]
})

inventory_expiry
```

```python
print("⚠️ Expiry Alert")
print()

for _, item in inventory_expiry.iterrows():

    if item['days_left'] <= 3:

        print(
            f"{item['ingredient']} should be used within "
            f"{item['days_left']} day(s)"
        )
```

```python
critical_items = len(
    inventory_expiry[
        inventory_expiry['days_left'] <= 3
    ]
)

total_items = len(inventory_expiry)

waste_score = (
    (total_items - critical_items)
    / total_items
) * 100

print(
    f"♻️ Food Waste Prevention Score: "
    f"{waste_score:.0f}/100"
)
```

## 5. Smart Shopping Prediction

```python
inventory_stock = pd.DataFrame({
    'ingredient': [
        'Eggs',
        'Milk',
        'Rice',
        'Chicken',
        'Tomato'
    ],

    'quantity': [
        2,
        1,
        10,
        1,
        3
    ]
})

inventory_stock
```

```python
def shopping_prediction(inventory_stock):

    print("🛒 Smart Shopping Assistant")
    print()

    low_stock = inventory_stock[
        inventory_stock['quantity'] <= 2
    ]

    for _, item in low_stock.iterrows():

        print(
            f"Add {item['ingredient']} "
            f"to your shopping list"
        )
```

```python
shopping_prediction(inventory_stock)
```

```python
import pandas as pd

recipes_real = pd.read_csv(
    "/kaggle/input/datasets/shuyangli94/food-com-recipes-and-user-interactions/RAW_recipes.csv"
)

print(recipes_real.shape)
recipes_real.head()
```

```python
recipes_real.columns
```

```python
recipes_real[['name', 'ingredients']].head()
```

```python
available_ingredients = [
    "ground beef",
    "yellow onions",
    "diced tomatoes"
]

matches = recipes_real[
    recipes_real["ingredients"].apply(
        lambda x: all(item in x for item in available_ingredients)
    )
]

matches[["name","ingredients"]].head(10)
```

```python
matches[
    [
        'name',
        'minutes',
        'n_steps',
        'ingredients',
        'steps'
    ]
].head(3)
```

```python
matches = matches.copy()

matches["score"] = 1000 - matches["minutes"]

matches = matches.sort_values(
    by="score",
    ascending=False
)

matches[
    ["name", "minutes", "n_steps"]
].head(5)
```

```python
user_profile = {
    "goal": "Weight Loss",
    "preferred_time": 60
}
```

```python
matches["time_score"] = (
    user_profile["preferred_time"]
    - abs(matches["minutes"] - user_profile["preferred_time"])
)

matches = matches.sort_values(
    by="time_score",
    ascending=False
)

matches[
    ["name", "minutes", "time_score"]
].head(10)
```
